````python
from typing import Annotated

from langchain_anthropic import ChatAnthropic
from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages


class State(TypedDict):
    messages: Annotated[list, add_messages]


graph_builder = StateGraph(State)


llm = ChatAnthropic(model="claude-3-5-sonnet-20240620")
# Modification: tell the LLM which tools it can call
llm_with_tools = llm.bind_tools(tools)


def chatbot(state: State):
    return {"messages": [llm_with_tools.invoke(state["messages"])]}


graph_builder.add_node("chatbot", chatbot)
````

````python
def route_tools(
    state: State,
):
    """
    Use in the conditional_edge to route to the ToolNode if the last message
    has tool calls. Otherwise, route to the end.
    """
    if isinstance(state, list):
        ai_message = state[-1]
    elif messages := state.get("messages", []):
        ai_message = messages[-1]
    else:
        raise ValueError(f"No messages found in input state to tool_edge: {state}")
    if hasattr(ai_message, "tool_calls") and len(ai_message.tool_calls) > 0:
        return "tools"
    return END


# The `tools_condition` function returns "tools" if the chatbot asks to use a tool, and "END" if
# it is fine directly responding. This conditional routing defines the main agent loop.
graph_builder.add_conditional_edges(
    "chatbot",
    route_tools,
    # The following dictionary lets you tell the graph to interpret the condition's outputs as a specific node
    # It defaults to the identity function, but if you
    # want to use a node named something else apart from "tools",
    # You can update the value of the dictionary to something else
    # e.g., "tools": "my_tools"
    {"tools": "tools", END: END},
)
# Any time a tool is called, we return to the chatbot to decide the next step
graph_builder.add_edge("tools", "chatbot")
graph_builder.add_edge(START, "chatbot")
graph = graph_builder.compile()
````

````python
import json

from typing import Annotated

from typing_extensions import TypedDict

from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langchain_anthropic import ChatAnthropic
from langgraph.graph import StateGraph
from langgraph.graph.message import add_messages

from langgraph.checkpoint.memory import MemorySaver
````

````python

# Define the initial state graph builder
graph_builder = StateGraph(State)

# Create an instance of the LLM
llm = ChatGroq(temperature=0, model_name="gpt-3.5-turbo")

def router(state: State):
    details = state["details"]
    if next:= details.get('intention' , None):
        if "sell" in next:
            return "selling"
        elif "buy" in next:
            return "buying"
        else:
            raise(" Something went wrong here at router")
    else:
        return END
    

def confirmation_agent(state: State):
    details = state['details']
    confirmation = details.get('details_confirmed')
    if not confirmation:
         state['details'] = 'Seccess'
         print("congrats! you have gained the plan for the stocks.")
    return
    # system = f"your are supposed to confirm the following information: {state['details']}\n\
    #     if confirm key has value none update the json with "
    # human = input("Hi im your selling agent.\n\
    #               Please selct any of the following plans:\
    #             A: $3,000 per share\
    #             B: $5,000 per share\
    #             C: $10,000 per share")
    # prompt = ChatPromptTemplate.from_messages([("system", system), ("human", human)])

    # chain = prompt | chat #| JsonOutputParser()
    # print("im at seller agent , messages: ",  state["messages"])
    # res = chain.invoke({"text": state["messages"]})
    # print("Seller says: -- ", res.content)

def seller(state: State):
    system = "you are a stocks seller agent, your job is to ask user which plan does he want to select from the plans below to sell his stocks:\
        A: $3,000 per share\
        B: $5,000 per share\
        C: $10,000 per share\
            If user provides a plan return it as a json/dict as key value pair"
    human = input("Hi im your selling agent.\n\
                  Please selct any of the following plans:\
                A: $3,000 per share\
                B: $5,000 per share\
                C: $10,000 per share")
    prompt = ChatPromptTemplate.from_messages([("system", system), ("human", human)])

    chain = prompt | chat #| JsonOutputParser()
    print("im at seller agent , messages: ",  state["messages"])
    res = chain.invoke({"text": state["messages"]})
    print("Seller says: -- ", res.content)
    try:
    # json_final_res = res.content | JsonOutputParser()
        cleaned_json_string = res.content.strip('```json').strip('```').strip("\n")
        json_final_res = json.loads(cleaned_json_string)
        print(json_final_res)
    except:
        print("Not a json this time...")
        print(res)
    else:
    
    # Extract the current messages and tasks
    # messages = state["messages"]
        details = state["details"]
        details['plan'] = json_final_res.get('plan')
    # Return the updated state with the new messages and modified tasks
        print(details)
        return {
            "messages": [res],
            "details": details,
        }



def buyer(state: State):
    system = "you are a stocks buyer agent, your job is to ask user which plan does he want to select from the plans below to buy stocks:\
        X: $3,000 per share\
        Y: $5,000 per share\
        Z: $10,000 per share\
            If user provides a plan return it as a json/dict as key value pair"
    human = input("Hi im your buyer agent.\n\
                Please selct any of the following plans:\
                X: $3,000 per share\
                Y: $5,000 per share\
                Z: $10,000 per share")
    prompt = ChatPromptTemplate.from_messages([("system", system), ("human", human)])

    chain = prompt | chat #| JsonOutputParser()
    print("im at buyer agent , messages: ",  state["messages"])
    res = chain.invoke({"text": state["messages"]})
    print("buyer says: -- ", res.content)
    try:
    # json_final_res = res.content | JsonOutputParser()
        cleaned_json_string = res.content.strip('```json').strip('```').strip("\n")
        json_final_res = json.loads(cleaned_json_string)
        print(json_final_res)
    except:
        print("Not a json this time...")
        print(res)
    else:
    
    # Extract the current messages and tasks
    # messages = state["messages"]
        details = state["details"]
        details['plan'] = json_final_res.get('plan')
    # Return the updated state with the new messages and modified tasks
        print(details)
        return {
            "messages": [res],
            "details": details,
        }



def chatbot(state: State):
    system = "You are a finance assistant. Your job is to ask user following information if he seems to be interusted in stocks: \
        user 'user_name', 'email' , 'phone' and his 'intention' to either buying or selling any stocks provided with its 'stock_symbol'. If user provides or has provided \
        these 4 details Return just the output in json format."
    
    # For each query always return a json with key 'stock' and value 'nasdaq'"
    
    # Yo uare provided with the json containing tasks. Your job is to perform the task(s), after asking user \
    #     relevent information and then place it against the task key as value. Return just the json after the updation."
    human = "User: {text}"
    prompt = ChatPromptTemplate.from_messages([("system", system), ("human", human)])

    chain = prompt | chat #| JsonOutputParser()

    res = chain.invoke({"text": state["messages"]}) #{"text": "hi i want to invest in stocks."}
    try:
    # json_final_res = res.content | JsonOutputParser()
        cleaned_json_string = res.content.strip('```json').strip('```').strip("\n")
        json_final_res = json.loads(cleaned_json_string)
        print(json_final_res)
    except:
        print("Not a json this time...")
        print(res)
    else:
    
    # Extract the current messages and tasks
    # messages = state["messages"]
        details = state["details"]
        details['name'] = json_final_res.get('user_name')
        details['email'] = json_final_res.get('email')
        details['phone'] = json_final_res.get('phone')
        details['intention'] = json_final_res.get('intention')
        details['stock_symbol'] = json_final_res.get('stock_symbol')
   
    # Return the updated state with the new messages and modified tasks
        print(details)
        return {
            "messages": [res],
            "details": details,
        }
graph_builder.add_conditional_edges(
    "chatbot",
    router,
    # The following dictionary lets you tell the graph to interpret the condition's outputs as a specific node
    # It defaults to the identity function, but if you
    # want to use a node named something else apart from "tools",
    # You can update the value of the dictionary to something else
    # e.g., "tools": "my_tools"
    {
        "selling": "seller",
        "buying": "buyer",
          END: END
    }, then= "confirmation_agent"
)

# Add the chatbot node and set entry/finish points
graph_builder.add_node("chatbot", chatbot)
graph_builder.add_node("seller", seller)
graph_builder.add_node("buyer", buyer)
graph_builder.add_node('confirmation_agent', confirmation_agent)
graph_builder.set_entry_point("chatbot")
graph_builder.set_finish_point("confirmation_agent")
graph = graph_builder.compile()

t1 = "hi i want to invest in stocks"
t2 = "sure my name is asif, asif@gmail.com and +9876543456 and im interusted in buying and my stock symbol is ltry"
# Example input
initial_state = {
    "messages": [("user", t2)],
    "details": {
        "name": None,
        "email": None,
        "phone": None,
        "intention": None,

        "stock_symbol": None,

        "details_confirmed": None,
    }
}
````
